### Eleminating Invalid Records

```python
invalid_condition = (df_cluster['prs_opened'] == 0) & \
                  ((df_cluster['prs_closed'] > 0) | (df_cluster['prs_merged'] > 0))

num_invalid = invalid_condition.sum()
print(num_invalid)
df_cluster  = df_cluster[~invalid_condition]
```

### Deriving Usefull Columns

```python
import numpy as np

df_cluster['pr_merge_rate'] = np.divide(df_cluster['prs_merged'], df_cluster['prs_opened'], 
                                        out=np.zeros_like(df_cluster['prs_merged'], dtype=float), 
                                        where=df_cluster['prs_opened']!=0)

df_cluster['prs_closed_unmerged'] = df_cluster['prs_closed'] - df_cluster['prs_merged']

df_cluster['pr_issue_ratio'] = np.divide(df_cluster['prs_opened'], df_cluster['issues_opened'],
                                         out=np.zeros_like(df_cluster['prs_opened'], dtype=float),
                                         where=df_cluster['issues_opened']!=0)

df_cluster['total_issue_activity'] = df_cluster['issues_opened'] + df_cluster['issues_closed']

df_cluster[['pr_merge_rate', 'prs_closed_unmerged', 'pr_issue_ratio', 'total_issue_activity']].head()
```

#### pr_merge_rate
This shows the percentage of a user's pull requests that are successfully accepted into the project. It's a key indicator of the quality and relevance of their contributions.

#### prs_closed_unmerged
This feature counts the number of pull requests that were closed without being merged. It represents the total volume of a user's work that was either rejected, withdrawn, or became obsolete.

#### pr_issue_ratio
This ratio compares if a user submits more solutions (PRs) than they report problems (issues). It helps distinguish active code contributors from users who primarily find and report bugs.

#### total_issue_activity
This is the sum of all issues a user has opened and closed. It serves as a straightforward measure of their overall engagement level with the project's issue tracking system.
